# 1046. Last Stone Weight


i have a list [2,7,4,1,8,1]

first = needs to iterate first largest element
second = needs to iterate second largest element

compare first and second amnd solves the condition


```python
import heapq
class Solution:
    def lastStoneWeight(self, stones:list[int]) -> int:

        stones = [-stone for stone in stones]
        heapq.heapify(stones)

        while len(stones)>1:
            first = -heapq.heappop(stones)
            second = -heapq.heappop(stones)

            if (first != second):
                heapq.heappush(stones, -(first-second))

        return -stones[0] if stones else 0

obj=Solution()

print(obj.lastStoneWeight([2,7,4,1,8,1]))


```

first thing is kth elemnet say for example 3rd largest element in list,

while we keep on adding element in list, we still needs to find kth largest

```python
import heapq
class kthlargest:
    def __init__(self,k:int,nums:list[int]):
        self.k=k
        self.nums=nums

        heapq.heapify(self.nums)

        while len(self.nums)>self.k:
            heapq.heappop(self.nums)

    def add(self, val: int) -> int:
        heapq.heappush(self.nums,val)

        if len(self.nums)>self.k:
            heapq.heappop(self.nums)

        return self.nums[0]

obj=kthlargest(3,[4,5,8,2])
print (obj.add(3))

```
